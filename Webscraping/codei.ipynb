```python
import requests 
from bs4 import BeautifulSoup
```

```python
product='iQOO Z10'
```

```python
search_url='https://www.amazon.in/s?k=iqooz10'
print(search_url)
```

```python
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/117.0.0.0 Safari/537.36"
    )
}

# Send request
response = requests.get(search_url,headers=headers)

# Check response
print("Status Code:", response.status_code)
```

```python
print(response.text[:1000])
```

```python


soup = BeautifulSoup(response.text, "html.parser")
```

```python
first_product = soup.find("div", {"data-component-type": "s-search-result"})
```

```python
if first_product:
    product_link_tag = first_product.find("a", {"class": "a-link-normal s-no-outline"})
    if product_link_tag:
        product_link = "https://www.amazon.com" + product_link_tag["href"]
        print("First product link:", product_link)
    else:
        print("Product link tag not found.")
else:
    print("No product found on the page.")
```
